```python
from scipy.stats import binom

n = 10
k = 8
p = 0.75

# pmf stands for Probability Mass Function (exact probability)
result = binom.pmf(k, n, p)

print(f"Probability of exactly {k} successes: {result:.4f}")
```

```python
from scipy.stats import binom

n = 10     # Total trials
p = 0.75    # Probability of success
y = 8      # We want P(Y >= 8)

# Approach A: 1 minus the CDF of 7 (the cumulative probability up to 7)
prob_cdf = 1 - binom.cdf(y - 1, n, p)

# Approach B: Survival Function (SF) - automatically calculates 1 - CDF
prob_sf = binom.sf(y - 1, n, p)

print(f"P(Y >= 8) using CDF: {prob_cdf:.4f}")
print(f"P(Y >= 8) using SF:  {prob_sf:.4f}")
```
